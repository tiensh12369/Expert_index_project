```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
import os
import expert_index_lib as eil

def compare_and_plot_top_authors(comparison_data, expert_index_data, data_dir):
    # Create a table to compare top authors based on different indices
    top_100_h_index = comparison_data.sort_values(by='h_index', ascending=False)
    top_100_g_index = comparison_data.sort_values(by='g_index', ascending=False)
    top_100_i10_index = comparison_data.sort_values(by='i10_index', ascending=False)
    top_100_expert_index = expert_index_data.sort_values(by='expert_index', ascending=False)[:77172]
    
    print(len(top_100_h_index), len(top_100_g_index), len(top_100_i10_index))
    
    print(top_100_h_index)
    print(top_100_g_index)
    print(top_100_i10_index)
    # Combine all top 100 lists into a single DataFrame
    rank_comparison_df = pd.DataFrame({
        'Rank': range(1, 52145),
        'Top 100 by h-index': top_100_h_index['author_id'].tolist(),
        'Top 100 by g-index': top_100_g_index['author_id'].tolist(),
        'Top 100 by i10-index': top_100_i10_index['author_id'].tolist()
        # 'Top 100 by expert-index': top_100_expert_index['author_id'].tolist()
    })

    print("Rank Comparison Table:")
    print(rank_comparison_df)
    rank_comparison_df.to_csv('rank_comparison_table_2012.csv', index=False)

    top_experts = expert_index_data.head(20)  # Select top 20 authors by expert index
    merged_top_data = pd.merge(top_experts, comparison_data, on="author_id")

    # Add paper_count and total_citations for each author, filtered by the publication year range
    author_paper_counts = []
    for author in top_experts['author_id']:
        # Load publications filtered by the year range 2009 to 2013
        publications = eil.load_author_publications(data_dir, author)
        
        # Filter publications between 2009 and 2013
        filtered_publications = publications[(publications['year'] >= 2009) & (publications['year'] <= 2013)]
        
        paper_count = len(filtered_publications)
        total_citations = filtered_publications['citations'].sum() if not filtered_publications.empty else 0
        
        author_paper_counts.append({
            'author_id': author, 
            'paper_count': paper_count, 
            'total_citations': total_citations
        })

    paper_count_df = pd.DataFrame(author_paper_counts)

    # Merge paper count and total_citations data
    merged_top_data = pd.merge(merged_top_data, paper_count_df, on='author_id')
     # Plot comparison of top experts with other indices
    plt.figure(figsize=(18, 8))

    plt.subplot(2, 3, 1)
    sns.barplot(x="author_id", y="h_index", data=merged_top_data, palette="viridis")
    plt.xlabel("Author ID")
    plt.ylabel("h-index")
    plt.title("Top Experts by Proposed Index vs h-index")
    plt.xticks(rotation=45)

    plt.subplot(2, 3, 2)
    sns.barplot(x="author_id", y="g_index", data=merged_top_data, palette="viridis")
    plt.xlabel("Author ID")
    plt.ylabel("g-index")
    plt.title("Top Experts by Proposed Index vs g-index")
    plt.xticks(rotation=45)

    plt.subplot(2, 3, 3)
    sns.barplot(x="author_id", y="i10_index", data=merged_top_data, palette="viridis")
    plt.xlabel("Author ID")
    plt.ylabel("i10-index")
    plt.title("Top Experts by Proposed Index vs i10-index")
    plt.xticks(rotation=45)

    # Plot number of papers and total citations for top experts
    plt.subplot(2, 3, 4)
    sns.barplot(x="author_id", y="paper_count", data=merged_top_data, palette="viridis")
    plt.xlabel("Author ID")
    plt.ylabel("Number of Papers")
    plt.title("Top Experts by Proposed Index vs Number of Papers")
    plt.xticks(rotation=45)
    plt.xlabel("Author ID")
    plt.ylabel("Number of Papers")
    plt.title("Top Experts by Proposed Index vs Number of Papers")
    plt.xticks(rotation=45)

    plt.subplot(2, 3, 5)
    sns.barplot(x="author_id", y="total_citations", data=merged_top_data, palette="viridis")
    plt.xlabel("Author ID")
    plt.ylabel("Total Citations")
    plt.title("Top Experts by Proposed Index vs Total Citations")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()

    plt.tight_layout()
    plt.show()
    
    return top_experts



def load_author_publications(data_dir, author_id, current_year=2013):
    if author_id is None:
        return pd.DataFrame(columns=["citations", "year"])
    
    author_file = os.path.join(data_dir, f"{author_id}_.dat")
    if not os.path.exists(author_file):
        print(f"File {author_file} does not exist.")
        return pd.DataFrame(columns=["citations", "year"])
    publications = []
    try:
        with open(author_file, 'r', encoding='cp1252') as f:
            for line in f:
                citations, year = line.strip().split('|')
                if int(year) >= 1973 and int(year) <= current_year:
                    publications.append((int(citations), int(year)))
    except UnicodeDecodeError:
        print(f"Failed to open file {author_file} with the given encoding.")
    except FileNotFoundError:
        print(f"Publication data for author {author_id} not found.")
    return pd.DataFrame(publications, columns=["citations", "year"])

# Experiment evaluation: Comparing proposed index with existing indices and calculating yearly stats
def evaluate_expert_indices_and_yearly_stats(authors_data, data_dir):
    # Comparison data and yearly statistics
    comparison_data = []
    yearly_stats = []
    
    for author in authors_data.itertuples():
        if author.author_id is None:
            continue
        publications = load_author_publications(data_dir, author.author_id)
        
        # 출판 연도가 2009년부터 2013년 사이인 논문만 필터링
        filtered_publications = publications[(publications['year'] >= 2012) & (publications['year'] <= 2013)]
        
        if filtered_publications.empty:
            continue
        
        # Calculate h-index
        sorted_citations = sorted(filtered_publications["citations"], reverse=True)
        h_index = sum(1 for i, c in enumerate(sorted_citations) if c >= i + 1)
        
        # Calculate g-index
        cumulative_citations = np.cumsum(sorted_citations)
        g_index = sum(1 for i, c in enumerate(cumulative_citations) if c >= (i + 1) ** 2)
        
        # Calculate i10-index
        i10_index = sum(1 for c in sorted_citations if c >= 10)
        
        comparison_data.append((author.author_id, h_index, g_index, i10_index))
        
        # 연도별 논문 수 및 인용 수 계산
        yearly_publications = filtered_publications.groupby('year').size().reset_index(name='paper_count')
        yearly_citations = filtered_publications.groupby('year')['citations'].sum().reset_index(name='total_citations')
        
        # 두 개의 데이터를 결합하여 하나의 테이블로
        yearly_stats_author = pd.merge(yearly_publications, yearly_citations, on='year')
        yearly_stats_author['author_id'] = author.author_id  # 각 데이터에 author_id 추가
        
        yearly_stats.append(yearly_stats_author)

    # Create DataFrames for comparison data and yearly statistics
    comparison_df = pd.DataFrame(comparison_data, columns=["author_id", "h_index", "g_index", "i10_index"])
    
    # 연도별 통계 데이터를 하나의 DataFrame으로 병합
    yearly_stats_df = pd.concat(yearly_stats, ignore_index=True)
    
    return comparison_df, yearly_stats_df


authors_file = r"./gsc_data/authors.all"
data_dir = r"./gsc_data/DATA/"
a = 50
b = 50

expert_index_filename = f"expert_index_results__2012.csv"
comparison_index_filename = "comparison_index_results_2012.csv"

# comparison_index_filename = f"comparison_index_results_alpha_{a}_beta_{b}.csv"

# Load data
authors_data = eil.load_authors_data(authors_file)

expert_index_data, comparison_data = eil.load_or_save_expert_index_data(authors_data, data_dir, expert_index_filename, comparison_index_filename, a, b)

# Plot results
eil.plot_expert_indices(expert_index_data, comparison_data)
eil.plot_expert_indices2(expert_index_data, comparison_data)

# Create a table to compare top 10 authors based on different indices
top_experts = compare_and_plot_top_authors(comparison_data, expert_index_data, data_dir)
```
